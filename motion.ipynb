```python
import serial
import serial.tools.list_ports
import binascii

def list_serial_ports():
    ports = serial.tools.list_ports.comports()
    print("Available serial ports:")
    for i, p in enumerate(ports):
        print(f"  {i}: {p.device}  ({p.description})")
    return ports

def main():
    ports = list_serial_ports()
    if not ports:
        print("No serial ports found.")
        return

    choice = input("Select port index: ").strip()
    if not choice.isdigit() or int(choice) >= len(ports):
        print("Invalid selection.")
        return

    port_name = ports[int(choice)].device
    print(f"Opening {port_name}…")

    ser = serial.Serial(port_name, 115200, timeout=0.2)

    print("Ready. Enter hex bytes (e.g. '01 02 00 10 12 C1').")
    print("Type 'exit' to quit.\n")

    while True:
        cmd = input("hex> ").strip()
        if cmd.lower() == "exit":
            break

        # strip spaces and convert to raw bytes
        cmd_clean = cmd.replace(" ", "")
        if not cmd_clean:
            continue

        try:
            ser.write(binascii.unhexlify(cmd_clean))
        except Exception as e:
            print(f"Error: {e}")
            continue

        data = ser.read(256)
        if data:
            print("recv:", data.hex(" "))
        else:
            print("recv: <no response>")

    ser.close()

if __name__ == "__main__":
    main()
```
